# Linked list

```python
class Node:
    def __init__(self,data):
        self.data=data
        self.next=None
        
a=Node(5)
b=Node(3)
c=Node(6)
d=Node(7)
a.next=b
b.next=c
c.next=d

head=a
print(head.next.data)
```

##### Iterate through a linked list

```python
curr=head
def linked_list(curr):
    while curr!=None:
        print(curr.data,end=" ")
        curr=curr.next
        
linked_list(curr)
```

##### insert a node at the beginning

```python
new_node=Node(1)
new_node.next=head
head=new_node
```

```python
linked_list(head)
```

##### Insert a node at the end

```python
last_node=Node(9)
last_node.next=None
curr=head
while curr.next!=None:
    curr=curr.next

curr.next=last_node
```

```python
linked_list(head)
```

##### Insert a node at k th position

```python
k=3
curr=head
new_node=Node(4)

for i in range(k-1):
    curr=curr.next

new_node.next=curr.next
curr.next=new_node
```

```python
linked_list(head)
```

##### Delete the first node

```python
head=head.next
linked_list(head)
```

##### Delete the last node

```python
curr=head
while curr.next.next!=None:
    curr=curr.next
curr.next=None
```

```python
linked_list(head)
```

##### Delete the k th node

```python
k=2
curr=head
for i in range(k-1):
    curr=curr.next

curr.next=curr.next.next
```

```python
linked_list(head)
```

##### Fast and slow concept in singly linked list

if we have to find the middle value of the linked list we can use this concept

```python
slow=head
fast=head

while fast!=None and fast.next!=None:
    slow=slow.next
    fast=fast.next.next
print(slow.data)
```

**Reverse a linked list**

```python
linked_list(head)
```

```python
def reverse_list(head):
    curr=head
    nxt=None
    prev=None
    
    while curr!=None:
        nxt=curr.next
        curr.next=prev
        prev=curr
        curr=nxt
    return prev

linked_list(reverse_list(head))
```
