### What is Generator?

**A generator is a special type of iterator that produces values lazily using the yield keyword.**

```python
def numbers():
    yield 1
    yield 2
    yield 3
    
gen=numbers()
print(type(gen))
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
print(next(gen))
```

```python
def demo():
    print("A")
    yield 1
    
    print("B")
    yield 2
    
    print("C")
    
gen=demo()
next(gen)
```

```python
next(gen)
```

```python
next(gen)
```

```python
square=(i*i for i in range(5))
type(square)
```

```python
next(square)
```

```python
next(square)
```

```python
next(square)
```

```python
next(square)
```

### Lazy Evaluation

**Lazy evaluation means computing values only when they are needed.**

```python
def numbers():
    print('Generating 1')
    yield 1
    
    print('Generating 2')
    yield 2
    
gen=numbers()
```

```python
print(next(gen))
```

```python
print(next(gen))
```

### Infinite Generator

```python
def counter():
    n=1
    while True:
        yield n
        n+=1
gen=counter()
```

```python
print(next(gen))
```

```python
print(next(gen))
```

| `return`          | `yield`                            |
| ----------------- | ---------------------------------- |
| Ends the function | Pauses the function                |
| Returns one value | Produces multiple values over time |
| State is lost     | State is preserved                 |


| Feature    | List      | Generator           |
| ---------- | --------- | ------------------- |
| Memory     | High      | Low                 |
| Evaluation | Immediate | Lazy                |
| Reusable   | Yes       | No (once exhausted) |
| Syntax     | `[]`      | `()`                |
| Type       | `list`    | `generator`         |
