# What is Variable?

**A variable is a name (reference) bound to an object in memory. Variables do not contain objects; they reference them.**

```python
x=10
```

# What is Object?

**Everything is an object**

ex:--10, 'hello', 5.6, True, {}

Everything has:

- Identity
- Type
- Value

```python
x = 10

print(id(x))
print(type(x))
print(x)
```

# What does id() return?

**id() returns the identity of an object. In CPython, it is typically the memory address where the object resides, although the language specification only guarantees that it is unique for the object's lifetime.**

# Mutable vs Immutable

**Immutable Objects**

Cannot be modified after creation.

Examples: 
- int
- float
- bool
- str
- tuple
- frozenset
- bytes

```python
x = "hello"

x = x + " world"
print(x)
```

Python does not modify the original string.

Instead:

old object

"hello"

↓

new object

"hello world"

**Mutable Objects**

Can change without creating a new object.

Examples:

- list
- dict
- set
- bytearray

```python
nums = [1,2]

nums.append(3)

print(nums)
```
