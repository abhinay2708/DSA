# What is Operator

**An operator in Python is a special symbol or keyword used to perform computations and operations on variables and values. The values that an operator acts upon are called operands. For example, in the expression 10 + 5, the + symbol is the operator, while 10 and 5 are the operands.**

**Python supports arithmetic, assignment, comparison, logical, identity, membership, and bitwise operators. Each category serves a different purpose in expressions and control flow.**

### Arithmetic Operators

| Operator | Meaning        |
| -------- | -------------- |
| `+`      | Addition       |
| `-`      | Subtraction    |
| `*`      | Multiplication |
| `/`      | Division       |
| `//`     | Floor Division |
| `%`      | Modulus        |
| `**`     | Exponentiation |

```python
a = 10
b = 3

print(a + b)    # 13
print(a - b)    # 7
print(a * b)    # 30
print(a / b)    # 3.333...
print(a // b)   # 3
print(a % b)    # 1
print(a ** b)   # 1000
```

```python
print(10/3) 
print(10//3) # floor division
```

```python
print(-10//3)
```

### Assignment Operator

### Comparison Operator

| Operator | Meaning               |
| -------- | --------------------- |
| `==`     | Equal                 |
| `!=`     | Not Equal             |
| `>`      | Greater Than          |
| `<`      | Less Than             |
| `>=`     | Greater Than or Equal |
| `<=`     | Less Than or Equal    |

```python
x=10

print(x==10)
print(x!=10)
print(x>5)
print(x<10)
print(x>=10)
print(x<=100)
```

```python
# Chain compariosn

print(1<=x<=10)
```

#### What is short-circuit evaluation?

**Python stops evaluating a logical expression as soon as the final result is determined.**

```python
False and print("Hello")
```

```python
True or print("Hello")
```

### Identity Operator

- is
- is not

```python
a = [1,2]
b = a

print(a is b)
print(a is not b)
```

### Membership Operator

- in 
- not in

```python
nums = [1,2,3]

print(2 in nums)
print(2 not in nums)
```
