# File Handling

### What is a File?

**A file is a named location on secondary storage used to permanently store data.**

old method--open(), close(), etc.

| Mode   | Meaning      | File Must Exist?       |
| ------ | ------------ | ---------------------- |
| `"r"`  | Read         | Yes                    |
| `"w"`  | Write        | No                     |
| `"a"`  | Append       | No                     |
| `"x"`  | Create       | Must not already exist |
| `"rb"` | Read binary  | Yes                    |
| `"wb"` | Write binary | No                     |


**opening a file**

```python
file=open("data.txt",'r')
```

new method-- with open()

```python
# Reading a file
with open("data.txt",'r') as f:
    content=f.read()
print(content)
```

```python
# Writing in a file

with open("data.txt",'w') as f:
    f.write("I am from WB\n")
```

```python
# appending in a file
with open("data.txt", 'a') as f:
    f.write("\nI am Abhinay")
```

```python
# Create a new file
with open("data1.txt",'x') as f:
    f.write("Hi i am Abhinay")
```

```python
# read
with open("data.txt",'r') as f:
    content=f.read()
print(content)
```

```python
# readline
with open("data.txt",'r') as f:
    content=f.readline()
print(content)
```

```python
# readlines
with open("data.txt",'r') as f:
    content=f.readlines()
print(content)
```

| Method / Mode | Purpose                       |
| ------------- | ----------------------------- |
| `"r"`         | Read                          |
| `"w"`         | Write (overwrite)             |
| `"a"`         | Append                        |
| `"x"`         | Create new file               |
| `read()`      | Read entire file              |
| `readline()`  | Read one line                 |
| `readlines()` | Read all lines into a list    |
| `with open()` | Automatic resource management |


### What is a csv file

**CSV (Comma-Separated Values) is a plain text file format where each line represents a row and commas separate the values.**

```python
# reading a csv file
import csv

with open("students.csv",'r') as f:
    reader=csv.reader(f)
    
    for row in reader:
        print(row)
```

```python
# Writing in a csv file

rows=[
    ['Name','Age','City'],
    ['Abhinay',24,'HYD'],
    ['Rahul',25,'UP']
]

with open("student.csv",'w') as f:
    writer=csv.writer(f)
    writer.writerows(rows)
```

### What is JSON?

**JSON (JavaScript Object Notation) is a lightweight text format used for storing and exchanging structured data.**

**Reading json**

```python
import json

with open("config.json",'r') as f:
    content=json.load(f)
print(content)
```

```python
# writing json
config={
    "epochs":10,
    "batch_size":32
}

with open("config.json",'w') as f:
    json.dump(config,f,indent=4)
```

```python
# saving a model in pockle file
import pickle
model="LinearRegression"
with open("model.pkl",'wb') as f:
    pickle.dump(model,file)
```

```python
# Loading a model

with open("model.pkl",'rb') as f:
    model=pickle.load(f)
```

| Tool      | Purpose                         |
