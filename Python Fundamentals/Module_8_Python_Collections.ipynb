## List

**A Python list is an ordered, mutable, dynamically sized sequence that can store references to objects of different types.**

**Ordered**

- Elements maintain their position.

```python
nums = [10, 20, 30]
print(nums)
```

**Mutable**

- This list can be change after cration.

```python
nums[0] = 100
print(nums)
```

**Dynamically Sized**

- You do not declare the size beforehand.

```python
nums = []

nums.append(10)
nums.append(20)
nums.append(30)

print(nums)
```

**Heterogeneous**

- A list can contain different object types.

```python
data = [10, "Python", 3.14, True]
print(data)
```

#### sort() vs sorted()

**list.sort() sorts the list in place and returns None, while sorted() returns a new sorted list and works with any iterable.**

```python
nums = [3, 1, 2]

result = nums.sort()
print(result)
```

```python
print(nums)
```

```python
nums = [3, 1, 2]

result = sorted(nums)
print(result)
```

```python
print(nums)
```

#### Reversing a List

**reverse()**

- Modifies the original list.

```python
nums=[1,2,3,4,5]
nums.reverse()
print(nums)
```

**slicing**

- Creates a new list.

```python
reversed_nums = nums[::-1]
print(reversed_nums)
```

**reversed()**

- This returns a reverse iterator, not a list.

```python
result = reversed(nums)
print(result)
```

```python
result = list(reversed(nums))
print(result)
```

## Tuple

**A tuple is an ordered, immutable sequence of object references.**

**The three important properties are:**

- Ordered → elements maintain position.
- Indexed → elements can be accessed using indexes.
- Immutable → the tuple structure cannot be modified after creation.

**Creating Tuple**

```python
t1=()
type(t1)
```

```python
nums=tuple()
print(nums)
```

```python
nums = (10, 20, 30)
print(nums)
```

```python
nums=tuple([1,2,3,4])
print(nums)
```

```python
nums=tuple({12,334,5,6})
print(nums)
```

**Tuples are Immutable**

```python
nums = (10, 20, 30)

nums[0] = 100
```

**What Does Tuple Immutability Actually Mean?**

Tuple immutability means the tuple's element references cannot be added, removed, or replaced after creation. However, if a tuple contains a mutable object, that nested object's internal state can still be modified.

```python
data = (10, [20, 30])
data[0] = 100
```

```python
data = (10, [20, 30])
data[1].append(40)
print(data)
```

**List vs Tuple**

| Feature               | List                | Tuple        |
| --------------------- | ------------------- | ------------ |
| Syntax                | `[1, 2]`            | `(1, 2)`     |
| Ordered               | Yes                 | Yes          |
| Indexed               | Yes                 | Yes          |
| Mutable               | Yes                 | No           |
| `append()`            | Yes                 | No           |
| `remove()`            | Yes                 | No           |
| Can be dictionary key | No                  | Sometimes    |
| Typical use           | Changing collection | Fixed record |


**Time Complexity**

| Operation    | Complexity |

| `t[a:b]`     |       O(k) |


## Set

**A set is a mutable, unordered collection of unique hashable elements, implemented using a hash table.**

**Every word matters:**

- Mutable → elements can be added and removed.
- Unordered → do not rely on positional order.
- Unique → duplicates are automatically removed.
- Hashable elements only → mutable objects such as lists cannot be set elements.

**Creating a set**

```python
nums=set()
print(nums)
```

```python
nums={10,20,30}
print(nums)
type(nums)
```

**Set do not support indexing**

```python
nums = {10, 20, 30}

print(nums[0])
```

**Complexity comparison**

| Operation         |           List |           Set |
| ----------------- | -------------: | ------------: |
| `x in collection` |           O(n) |  O(1) average |
| Add at end / add  | O(1) amortized |  O(1) average |
| Remove by value   |           O(n) |  O(1) average |
| Index access      |           O(1) | Not supported |
| Duplicates        |        Allowed |   Not allowed |

## Dictionary(Hash Map)

**A dictionary is a mutable mapping of unique, hashable keys to values, implemented using a hash table.**

- Mutable → you can add, update, and remove entries.
- Mapping → it maps keys to values.
- Unique keys → duplicate keys are not allowed.
- Hashable keys → keys must be hashable.
- Hash table → provides fast average-case lookup.

## Comprehensions

#### List Comprehension

- [expression for item in iterable]

```python
nums = []

for i in range(5):
    nums.append(i * i)

print(nums)
```

```python
nums = [i * i for i in range(5)]

print(nums)
```

```python
names = [word.lower() for word in ["AI", "ML", "DL"]]
print(names)
```

```python
even = []

for i in range(10):
    if i % 2 == 0:
        even.append(i)

even
```

```python
even=[i for i in range(10) if i%2==0]
print(even)
```

```python
nums=[1,2,5,-3,0,-5,7,8,-1]
result = []

for x in nums:
    if x > 0:
        result.append(x)
    else:
        result.append(0)

print(result)
```

```python
result=[x if x>0 else 0 for x in nums]
print(result)
```

**Multiple Loops**

```python
for i in range(2):
    for j in range(3):
        print(i, j)
```

```python
pairs = [(i, j)
         for i in range(2)
         for j in range(3)]

print(pairs)
```

```python
matrix = [
    [1,2],
    [3,4]
]

result = []

for row in matrix:
    for value in row:
        result.append(value)
        
print(result)
```

```python
result = [value for row in matrix for value in row]
print(result)
```

#### Dictionary Comprehension

```python
square={}

for i in range(5):
    square[i]=i*i

print(square)
```

```python
square={i:i*i for i in range(5)}
print(square)
```

```python
# Dictionary with condition

square={i:i*i for i in range(10) if i%2==0}
print(square)
```

```python
# Inverting a Dictionary

student = {
    "A":1,
    "B":2
}

reverse={v:k for k,v in student.items()}
print(reverse)
```

#### Set Comprehension

```python
square = set()

for i in range(5):
    square.add(i*i)

print(square)
```

```python
square={i*i for i in range(5)}
print(square)
```

## enumerate & zip

- enumerate()
- zip()
- sorted()
- reversed()
- any()
- all()

**enumerate()**

```python
names = ["Alice", "Bob", "Charlie"]

for i in range(len(names)):
    print(i, names[i])
```

```python
for index,name in enumerate(names):
    print(index,name)
```

```python
# start index

for i, name in enumerate(names, start=1):
    print(i, name)
```

**zip()**

```python
names = ["Alice", "Bob", "Charlie"]

scores = [90, 85, 95]

for i in range(len(names)):
    print(names[i],scores[i])
```

```python
for name,score in zip(names,scores):
    print(name,score)
```

```python
# unzipping

pairs = [
    ("A",90),
    ("B",80)
]

names,scores=zip(*pairs)
print(names)
print(scores)
```

**sorted()**

- It returns a new list, original remian unchanged.

```python
nums = [3,1,2]
result = sorted(nums)
print(nums)
print(result)
```

```python
# Sorting Dictionaries

student = {
    "Bob":80,
    "Alice":95,
    "Charlie":70
}

sorted(student)
```

```python
sorted(student.items(),
       key=lambda x:x[1])
```
