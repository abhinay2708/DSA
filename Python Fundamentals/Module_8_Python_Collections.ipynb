## List

**A Python list is an ordered, mutable, dynamically sized sequence that can store references to objects of different types.**

**Ordered**

- Elements maintain their position.

```python
nums = [10, 20, 30]
print(nums)
```

**Mutable**

- This list can be change after cration.

```python
nums[0] = 100
print(nums)
```

**Dynamically Sized**

- You do not declare the size beforehand.

```python
nums = []

nums.append(10)
nums.append(20)
nums.append(30)

print(nums)
```

**Heterogeneous**

- A list can contain different object types.

```python
data = [10, "Python", 3.14, True]
print(data)
```

**append() vs extend()**

```python
a = [1, 2]

a.append([3, 4])
print(a)
```

The entire list [3, 4] is added as one element.

```python
a = [1, 2]

a.extend([3, 4])
print(a)
```

#### sort() vs sorted()

**list.sort() sorts the list in place and returns None, while sorted() returns a new sorted list and works with any iterable.**

```python
nums = [3, 1, 2]

result = nums.sort()
print(result)
```

```python
print(nums)
```

```python
nums = [3, 1, 2]

result = sorted(nums)
print(result)
```

```python
print(nums)
```

#### Reversing a List

**reverse()**

- Modifies the original list.

```python
nums=[1,2,3,4,5]
nums.reverse()
print(nums)
```

**slicing**

- Creates a new list.

```python
reversed_nums = nums[::-1]
print(reversed_nums)
```

**reversed()**

- This returns a reverse iterator, not a list.

```python
result = reversed(nums)
print(result)
```

```python
result = list(reversed(nums))
print(result)
```

### Essential List Time Complexities

| Operation           |     Complexity |
| ------------------- | -------------: |
| `nums[i]`           |           O(1) |
| `nums[i] = x`       |           O(1) |
| `append(x)`         | O(1) amortized |
| `pop()`             |           O(1) |
| `insert(0, x)`      |           O(n) |
| `pop(0)`            |           O(n) |
| `x in nums`         |           O(n) |
| `index(x)`          |           O(n) |
| `count(x)`          |           O(n) |
| `sort()`            |     O(n log n) |
| Slicing `nums[a:b]` |           O(k) |
| `len(nums)`         |           O(1) |


```python
nums = [3, 1, 2]

result = nums.sort()

print(nums)
print(result)
```

## Tuple

**A tuple is an ordered, immutable sequence of object references.**

**The three important properties are:**

- Ordered → elements maintain position.
- Indexed → elements can be accessed using indexes.
- Immutable → the tuple structure cannot be modified after creation.

**Creating Tuple**

```python
t1=()
type(t1)
```

```python
nums=tuple()
print(nums)
```

```python
nums = (10, 20, 30)
print(nums)
```

```python
nums=tuple([1,2,3,4])
print(nums)
```

```python
nums=tuple({12,334,5,6})
print(nums)
```

```python
nums=tuple({1:2,3:4,5:6})
print(nums)
```

**Create a single element tuple**

```python
t=(2,)
print(t)
type(t)
```

**The comma creates the tuple, not the parentheses.**

```python
t=1,
print(t)
type(t)
```

**Indexing**

```python
nums = (10, 20, 30, 40)

nums[0]
```

```python
nums[-1]
```

**Slicing**

```python
nums = (10, 20, 30, 40, 50)

print(nums[1:4])
```

```python
nums[::-1]
```

**Tuples are Immutable**

```python
nums = (10, 20, 30)

nums[0] = 100
```

**What Does Tuple Immutability Actually Mean?**

Tuple immutability means the tuple's element references cannot be added, removed, or replaced after creation. However, if a tuple contains a mutable object, that nested object's internal state can still be modified.

```python
data = (10, [20, 30])
data[0] = 100
```

```python
data = (10, [20, 30])
data[1].append(40)
print(data)
```

**Tuple Methods**

- count()
- index()

```python
nums = (1, 2, 2, 2, 3)

nums.count(2)
```

```python
nums.index(2)
```

**Tuple Packing**

Python can automatically pack multiple values into a tuple.

```python
person = "Abhinay", 25, "AI Engineer"
print(person)
```

**Tuple Unpacking**

```python
person = ("Abhinay", 25, "AI Engineer")
name, age, role = person
```

```python
print(name)
print(age)
print(role)
```

**Extended Unpacking**

```python
nums = (10, 20, 30, 40, 50)

first, *middle, last = nums
```

```python
print(first)
print(middle)
print(last)
```

**Functions and Tuples**

```python
def calculate(a, b):
    return a + b, a - b
result = calculate(10, 5)
print(result)
```

**List vs Tuple**

| Feature               | List                | Tuple        |
| --------------------- | ------------------- | ------------ |
| Syntax                | `[1, 2]`            | `(1, 2)`     |
| Ordered               | Yes                 | Yes          |
| Indexed               | Yes                 | Yes          |
| Mutable               | Yes                 | No           |
| `append()`            | Yes                 | No           |
| `remove()`            | Yes                 | No           |
| Can be dictionary key | No                  | Sometimes    |
| Typical use           | Changing collection | Fixed record |


**Time Complexity**

| Operation    | Complexity |
| ------------ | ---------: |
| `t[i]`       |       O(1) |
| `len(t)`     |       O(1) |
| `x in t`     |       O(n) |
| `t.index(x)` |       O(n) |
| `t.count(x)` |       O(n) |
| `t[a:b]`     |       O(k) |
