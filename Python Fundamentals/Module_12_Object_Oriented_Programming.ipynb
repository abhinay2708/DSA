# Object Oriented Programming

**Object-Oriented Programming (OOP) is a programming paradigm that organizes software using objects, where each object combines data (attributes) and behaviour (methods).**

### Class

**A class is a blueprint or template used to create objects.**

### Object

**An object is an instance of a class.**

### Attributes

**Attributes are variables belonging to an object.**

### Methods

**Methods are functions inside classes.**

### Self

**Self referes to the current object**

### Constructor(__init__)

**__init__ is the constructor method that is automatically executed when a new object is created.**

```python
class Student:
    pass
s1=Student()
```

```python
class Student:
    def greet(self):
        print("Good Morning")

s1=Student()
s1.greet()
```

```python
class Student:
    def __init__(self):
        print("Hello")
s1=Student()
```

```python
class Student:
    def __init__(self,name,age):
        self.age=age
        self.name=name
    
    def intro(self):
        print(f"Hello i am {self.name} and my age is {self.age}")

s1=Student("Abhinay",25)
s1.intro()
```

### Instance Variable

**Instance variables belong to a specific object. Each object has its own separate copy.**

```python
class Student:

    def __init__(self, name, age):
        self.name = name
        self.age = age
        
s1=Student("Abhinay", 100)
s2=Student("Rahul",200)
print(s1.name, s1.age)
print(s2.name, s2.age)
```

### Class Variable

**Class variable declared inside the class outside any method, and after creating the objects, any object can easily access it.**

```python
class Students:
    company="MedIntelX"
    
    def __init__(self,name,age):
        self.name=name
        self.age=age
    
s1=Students("Abhinay", 25)
s2=Students("Ravi", 26)

print(s1.company)
print(s2.company)
```

### Instance Variable vs Class Variable

| Instance Variable           | Class Variable                |
| --------------------------- | ----------------------------- |
| Belongs to an object        | Belongs to the class          |
| Separate copy per object    | Shared by all objects         |
| Usually defined with `self` | Defined directly in the class |


### Instance Method

**An instance method operates on a specific object and receives self as its first parameter.**

```python
class Student:
    def __init__(self,name):
        self.name=name
    
    def intro(self):
        print(self.name)

s1=Student("Abhinay")
s1.intro()
```

### Class Method

```python
class Student:

    college = "ABC"

    @classmethod
    def show_college(cls):
        print(cls.college)
        
s1=Student()
s1.show_college()
```

```python
class Student:

    college = "ABC"
    
    @classmethod
    def change_college(cls, new_name):
        cls.college = new_name

Student.change_college("XYZ")
Student.college
```

### Static Method

```python
class Math:

    @staticmethod
    def square(x):
        return x * x

Math.square(5)
```

| Method   | First Parameter | Access                         |
| -------- | --------------- | ------------------------------ |
| Instance | `self`          | Object data                    |
| Class    | `cls`           | Class data                     |
| Static   | None            | Neither object nor class state |


### Inheritence

**Inheritance is an OOP mechanism where one class acquires the attributes and methods of another class, allowing code reuse and extension.**

```python
class Animal:
    def eat(self):
        print("Eating")


class Dog(Animal):
    def bark(self):
        print("barking")

class Cat(Animal):
    pass

d=Dog()
c=Cat()

d.eat()
```

```python
c.eat()
```

#### super()

**super() returns a proxy object that allows a child class to call methods of its parent class.**

```python
class Animal:
    def __init__(self,name):
        self.name=name
    
class Dog(Animal):
    def __init__(self,name,breed):
        super().__init__(name)
        self.breed=breed

d=Dog("Tommy","pitbull")
print(d.name)
print(d.breed)
```

### Method Overriding

```python
class Animal:
    def sound(self):
        print("Animal sound")

class Dog(Animal):
    def sound(self):
        super().sound()
        print("Dog is barking")
        
d=Dog()
d.sound()
```

| Concept           | Meaning                             |
| ----------------- | ----------------------------------- |
| Inheritance       | Child acquires parent functionality |
| Parent Class      | Base class                          |
| Child Class       | Derived class                       |
| `super()`         | Access parent methods               |
| Method Overriding | Replace parent implementation       |
| MRO               | Search order for methods            |
| `isinstance()`    | Object relationship check           |
| `issubclass()`    | Class relationship check            |


### Polymorphism

**Polymorphism means one interface, many implementations. The same method name behaves differently depending on the object.**

```python
class Dog:
    def sound(self):
        print("Bhoo")

class Cat:
    def sound(self):
        print("Meow")

class Cow:
    def sound(self):
        print("Bowwww")

animals=[Dog(), Cat(), Cow()]

for animal in animals:
    animal.sound()
```

### Method Overloading

```python
class Calculation:
    def add(self,a,b):
        return a+b
    
    def add(self,a,b,c):
        return a+b+c
    
c=Calculation()
c.add(5,6,7)
```

### Encapsulation

**Encapsulation is the process of bundling data and methods together while restricting direct access to internal data.**

```python
# Protected member

class BankAccount:
    def __init__(self):
        self._balance=1000

bank=BankAccount()
bank._balance
```

```python
# Private member

class BankAccount:
    def __init__(self):
        self.__balance = 1000

bank=BankAccount()
bank.__balance
```

```python
# getter method
class BankAccount:

    def __init__(self):
        self.__balance = 1000
        
    def get_balance(self):
        return self.__balance

account=BankAccount()
account.get_balance()
```

```python
# setter method

class BankAccount:

    def __init__(self):
        self.__balance = 1000
    
    def set_balance(self, amount):
        if amount>0:
            self.__balance=amount
    
    def get_balance(self):
        return self.__balance

account=BankAccount()
print(account.get_balance())
account.set_balance(5000)
print(account.get_balance())
```

```python
# Properties

class BankAccount:
    def __init__(self):
        self.__balance=1000
    
    @property
    def balance(self):
        return self.__balance

account=BankAccount()
account.balance
```

```python
account._BankAccount__balance
```

### Abstraction

**Abstraction hides implementation details and exposes only essential functionality.**

```python
from abc import ABC, abstractmethod

class Animal(ABC):
    @abstractmethod
    def sound(self):
        pass

class Dog:
    def sound(self):
        print("Barking")

d=Dog()
d.sound()
```

| Pillar        | Purpose                             |
| ------------- | ----------------------------------- |
| Inheritance   | Reuse existing code                 |
| Polymorphism  | Same interface, different behaviour |
| Encapsulation | Protect internal data               |
| Abstraction   | Hide implementation details         |

| Concept           | Meaning                                   |
| ----------------- | ----------------------------------------- |
| Polymorphism      | Same interface, different implementations |
| Duck Typing       | Behavior matters more than type           |
| Encapsulation     | Bundle data and behavior, control access  |
| `_var`            | Protected convention                      |
| `__var`           | Name-mangled private attribute            |
| `@property`       | Pythonic getter/setter                    |
| Abstraction       | Expose only essential operations          |
| `ABC`             | Abstract Base Class                       |
| `@abstractmethod` | Must be implemented by subclasses         |
