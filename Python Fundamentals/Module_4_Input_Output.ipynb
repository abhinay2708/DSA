# Input & Output

1. Print function

```python
print("Hello")
```

It accept multiple arguments

```python
name="Abhinay"
age=25
print(name, age)
```

```python
# seperator parameter in print

print("Data","Science",sep="|")
```

```python
# end parameter(The default is \n)

print("Hello", end=" ")
print("World")
```

2. Input function

```python
# Bydefault it takes string and convert string 
name=input("Enter your name")
print(name)
```

```python
age=input()
print(age)
type(age)
```

3. Type conversion

```python
age=int(input("Enter your age"))
print(age)
type(age)
```

4. Multiple Inputs

```python
a,b=map(int,input().split())
print(a,b)
```

5. Reading a List

```python
l=list(map(int,input().split()))
print(l)
```

```python
t=(1,2,3,4,5)
ll=[i for i in t]
print(ll)
```

6. f-string

```python
name="Abhinay"
age=25
print(f"Hi my name is {name} and my age is {age}")
```

```python
# Expression inside f string
a=10
b=20
print(f"Sum of {a} and {b} is {a+b}")
```

```python
# Formatting numbers

pi=3.1415926574
print(f"{pi:.2f}")
```

```python
# Padding Numbers

x=7
print(f"{x:03}")
```

```python
# Allignment
#left
print(f"{'AI':<10} Engineer")
```

```python
# right
print(f"{'Ai':>10} engineer")
```

```python
print(f"AI {'Engineer':^10}")
```
