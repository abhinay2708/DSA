# Context Managers

**A context manager is an object that properly acquires and releases resources by implementing the context management protocol.**

```python
# Before with
file=open("data.txt","r")
data=file.read()
file.close()
```

```python
# Zero Division Error
file=open("data.txt","r")
data=file.read()
10/0
file.close()
```

```python
# old solution
file = open("data.txt")

try:
    data = file.read()
finally:
    file.close()
```

```python
#new solution
with open("data.txt", "r") as f:
    data=f.read()
```

### The context manager protocol

**The context manager must implement two methods:**

- __enter__()
- __exit__()

**__enter__()**

- __enter__() runs when entering the with block.

```python
class Demo:
    def __enter__(self):
        print("Entering")
        return self
    
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exiting")
        

with Demo():
    print("Inside")
```

```python
class Demo:
    def __enter__(self):
        print("Start")
        return self
    
    def hello(self):
        print("Hello")
        
    def __exit__(self,exc_type, exc_value, traceback):
        print("End")


with Demo() as d:
    d.hello()
```

**__exit__()**

- __exit__() is called whenever execution leaves the with block, even if an exception occurs.

```python
def __exit__(
        self,
        exc_type,
        exc_value,
        traceback
):
```

**Parameters**

| Parameter   | Meaning          |
| ----------- | ---------------- |
| `exc_type`  | Exception class  |
| `exc_value` | Exception object |
| `traceback` | Stack trace      |

```python
# Handling exception

class Demo:
    def __enter__(self):
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print(exc_type)
        return True
    
with Demo():
    10/0
```

```python
# Handling exception

class Demo:
    def __enter__(self):
        return self
    def __exit__(self, exc_type, exc_value, traceback):
        print(exc_type)
        return False
    
with Demo():
    10/0
```

```python
# Building my own context manager

import time

class Timer:
    def __enter__(self):
        self.start=time.time()
        return self
    
    def __exit__(self, exc_type, exc_value, traceback):
        print(time.time()-self.start)

with Timer():
    total=sum(range(1000000))
```

```python
class Demo:
    def __enter__(self):
        print("Entering")
        return self
    
    def greet(self):
        print("Good Morning")
    
    def __exit__(self, exc_type, exc_value, traceback):
        print("Exit")

with Demo() as d:
    d.greet()
```

### contextlib.contextmanager

**@contextmanager from contextlib allows us to create a context manager using a generator function, where code before yield performs setup and code after yield performs cleanup.**

```python
from contextlib import contextmanager

@contextmanager
def demo():
    print("start")
    yield
    print("end")

with demo():
    print("Hi")
```

```python
from contextlib import contextmanager

@contextmanager
def resource():
    print("Openning resource")
    yield
    print("Closing resource")

with resource():
    print("Using resource")
```

### Exception handling inside a context manager

```python
from contextlib import contextmanager

@contextmanager
def demo():
    print("Start")
    
    try:
        yield
        
    except ValueError:
        print("ValueError handled")
    
    finally:
        print("Cleanup")

with demo():
    raise ValueError("Invalid Error")
```

**Difference between Decorators and context manager**

| Decorator               | Context Manager      |
| ----------------------- | -------------------- |
| Wraps functions/classes | Manages a block      |
| Uses `@`                | Uses `with`          |
| Changes behaviour       | Manages lifecycle    |
| Common for logging/auth | Common for resources |
