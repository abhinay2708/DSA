# Control Flow

**Control flow is the order in which a program executes statements. It allows the program to make decisions and execute different blocks of code based on conditions.**

```python
age = 20

if age >= 18:
    print("Eligible")
else:
    print("Not Eligible")
```

1. If statement

```python
marks = 85

if marks >= 40:
    print("Pass")
```

2. If-Else

```python
age = 16

if age >= 18:
    print("Vote")
else:
    print("Cannot Vote")
```

3. If-Elif-Else

```python
marks = 78

if marks >= 90:
    print("A")

elif marks >= 80:
    print("B")

elif marks >= 70:
    print("C")

else:
    print("Fail")
```

4. Nested If

```python
age = 25
citizen = True

if age >= 18:
    if citizen:
        print("Eligible")
```

5. Ternery Operator

```python
age=25
status="Adult" if age>18 else "Minor"
print(status)
```

4. match case

```python
d=3
match d:
    case 1:
        print("Hi")
    case 2:
        print("Hello")
    
    case _:
        print("Oye")
        
```
