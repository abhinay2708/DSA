# Decorators

**A decorator is a function that takes another function as input, adds additional behaviour, and returns a new function without modifying the original function's source code.**

```python
def greet():
    print("Hello")
    
def say(func):
    func()

hi=say
say(greet)
```

```python
def outer():
    
    def inner():
        print("Hello")
    return inner
x=outer()
x()
```

```python
def outer():

    message = "Hello"

    def inner():
        print(message)

    return inner

func = outer()

func()
```

```python
def greet():
    print("Hello")
    
def outer(func):
    
    def inner():
        print("Before")
        
        func()
        
        print("After")
    return inner

new=outer(greet)
new()
```

| Concept               | Role in Decorators             |
| --------------------- | ------------------------------ |
| Function Object       | Can be passed around           |
| Higher-order Function | Accepts/returns functions      |
| Nested Function       | Builds the wrapper             |
| Closure               | Remembers the wrapped function |
| Decorator             | Returns an enhanced function   |


```python
def outer(func):
    def inner():
        print("Before")
        func()
        print("After")
    return inner

@outer
def greet():
    print("Hello")

greet()
```

```python
def decorator(func):
    def wraper(*args):
        print("Start")
        result=func(*args)
        print("End")
        return result
    return wraper

@decorator
def add(a,b):
    print(a+b)

add(5,6)
```

```python
def decorator(func):
    def wraper(*args,**kwargs):
        print("Start")
        result=func(*args,**kwargs)
        print("End")
        return result
    return wraper

@decorator
def add(a,b,c):
    print(a+b+c)

add(5,6,c=10)
```

```python
import time

def timer(func):
    def wraper(*args,**kwargs):
        start=time.time()
        print("Execution Started")
        result=func(*args,**kwargs)
        print("Execution finished")
        end=time.time()
        print(f"Total time= {end-start}")
        return result
    return wraper

@timer
def loop(a,b):
    s=0
    for i in range(a,b):
        s+=i
    print(f"The result is : {s}")

loop(1,1000000)
```

```python
# Logger

def logger(func):
    def wraper(*args,**kwargs):
        print(f"Calling {func.__name__}")
        result=func(*args,**kwargs)
        print(f"Ending {func.__name__}")
        return result
    return wraper

@logger
def add(a,b):
    print(a+b)
    
add(5,6)
```

```python
# The metadata problem
print(add.__name__)
```

```python
# Solution : functools.wraps

from functools import wraps

def logger(func):
    @wraps(func)
    def wraper(*args,**kwargs):
        print(f"Calling {func.__name__}")
        result=func(*args,**kwargs)
        print(f"Ending {func.__name__}")
        return result
    return wraper

@logger
def add(a,b):
    print(a+b)
    
add(5,6)
```

```python
print(add.__name__)
```

```python
def decor1(func):
    def wraper(*args,**kwargs):
        result=func(*args,**kwargs)
        print("decor1 executing")
        print("Doubling the result")
        return f"final answer {result*2}"
    return wraper

def decor2(func):
    def wraper(*args,**kwargs):
        result=func(*args,**kwargs)
        print("decor2 executing")
        print("Squaring the result")
        return result*result
    return wraper

@decor1
@decor2
def original(a):
    print(f"Original number is : {a}")
    return a

original(10)
```

| Feature             | Purpose               |
| ------------------- | --------------------- |
| `@decorator`        | Cleaner syntax        |
| `*args`             | Positional arguments  |
| `**kwargs`          | Keyword arguments     |
| `return result`     | Preserve return value |
| `@wraps`            | Preserve metadata     |
| Multiple decorators | Compose behaviour     |
