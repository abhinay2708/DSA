# Functions

**A function is a reusable block of code that performs a specific task. It can accept inputs (parameters), execute logic, and optionally return a value.**

```python
def greet():
    return "Hi Abhinay"
greet()
```

```python
def add(a,b):
    return a+b

add(5,6)
```

```python
add(100,1989)
```

#### Parameter vs Arguments

**A parameter is a variable defined in the function declaration. An argument is the actual value passed to the function when it is called.**

```python
def greet(name):
    return f"Hi {name}"

greet("Abhinay")
```

#### print() vs return

**print()**

- Displays output
- Used for debugging or user interaction
- Does not send a value back to the caller

**return**

- Ends function execution
- Sends a value back to the caller
- Can be stored or reused

```python
def add(a,b):
    print(a+b)
    
x=add(5,6)
print(x)
```

```python
def add(a,b):
    return a+b

x=add(5,6)
print(x)
```

```python
def calculate(a,b):
    return a+b,a-b

res=calculate(20,10)
print(res)
```

### Parameters and Arguments

##### Positional Arguments

**Python assigns arguments to parameters based on their position in the function call.**

```python
def introduce(name, age):
    print(name, age)

introduce("Abhinay", 25)
```

##### Keyword Arguments

```python
def introduce(name, age):
    print(name, age)

introduce(name="Abhinay", age=25)
```

##### Mixing Positional and Keyword Arguments

- Python first fills positional arguments, then keyword arguments.
- Positional arguments must come before keyword arguments.

```python
def introduce(name, age):
    print(name, age)
    
introduce(name="Abhinay",25)
```

```python
introduce("Abhinay",age=25)
```

##### Default Arguments

```python
def greet(name, city="Hyderabad"):
    print(name, city)
    
greet("Abhinay")
```

```python
greet("Abhinay", "Bangalore")
```

##### Mutable Default Arguments

**the default list is created once, when the function is defined—not each time it's called.
Every call without an explicit items argument uses the same list object.**

```python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(1))
```

```python
print(add_item(2))
```

##### Keyword only parameters

**The * means:
Parameters after * must be supplied using keywords.**

```python
def train(model, *, epochs, batch_size):
    print(model, epochs, batch_size)

train("CNN", epochs=10, batch_size=32)
```

```python
train("CNN", 10, 32)
```

##### Argument unpacking

```python
def add(*item):
    return sum(item)

add(1,2,3)
```

##### Dictionary Unpacking

```python
def introduce(name,age):
    print(name,age)

person={'name':"Abhinay",'age':25}
introduce(**person)
```

```python
def add(a, b, c):
    return a + b + c

nums = [10, 20, 30]

print(add(*nums))
```
