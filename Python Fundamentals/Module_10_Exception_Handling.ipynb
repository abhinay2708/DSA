# Exception Handling

##### What is an Exception?

**An exception is an event that occurs during program execution which interrupts the normal flow of the program.**

**Common Python Exception**

| Exception           | Cause                     |
| ------------------- | ------------------------- |
| `ZeroDivisionError` | Divide by zero            |
| `ValueError`        | Invalid value             |
| `TypeError`         | Wrong data type           |
| `IndexError`        | Invalid list index        |
| `KeyError`          | Missing dictionary key    |
| `AttributeError`    | Object lacks an attribute |
| `FileNotFoundError` | File does not exist       |
| `NameError`         | Variable not defined      |

**try**

The try block contains code that might raise an exception.

**except**

Handles a specific exception

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

```python
try:
    value = int(input())
    print(10 / value)

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

**else**

The else block executes only if no exception occurs.

```python
try:
    print(10 / 2)

except ZeroDivisionError:
    print("Error")

else:
    print("Everything succeeded")
```

**finally**

This block always execute

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Error")

finally:
    print("Cleanup")
```

**raise**

- The raise statement is used to explicitly generate an exception.

```python
def set_age(age):

    if age < 0:
        raise ValueError("Age cannot be negative.")

    return age

set_age(-5)
```

##### Built in Exception

| Exception           | When to Use                 |
| ------------------- | --------------------------- |
| `ValueError`        | Correct type, invalid value |
| `TypeError`         | Wrong data type             |
| `KeyError`          | Missing dictionary key      |
| `IndexError`        | Invalid index               |
| `FileNotFoundError` | Missing file                |
| `PermissionError`   | No access permission        |
| `RuntimeError`      | General runtime issue       |


```python
class AccountLockedError(Exception):
    pass

raise AccountLockedError("Account is locked.")
```

```python
class AbhinayPermissionError(Exception):
    pass
raise AbhinayPermissionError("Abhinay didn't give permission")
```

```python
class InvalidScoreError(Exception):
    pass
```

```python
def check(score):
    if score<0:
        raise InvalidScoreError("Score can not be negative")
    
```

```python
try:
    check(-10)
except InvalidScoreError as e:
    print(e)
```

```python
# Exception Chaining
try:
    int("abc")

except ValueError as e:
    raise RuntimeError("Invalid user input") from e
```

**assert**

- assert is used to verify assumptions during development.

```python
age = 20

assert age > 0
```

```python
age = -5

assert age > 0
```

```python
assert age > 0, "Age must be positive."
```

| Concept            | Purpose                          |
| ------------------ | -------------------------------- |
| `raise`            | Explicitly generate an exception |
| Custom Exception   | Represent domain-specific errors |
| `raise ... from e` | Preserve the original exception  |
| `assert`           | Verify development assumptions   |
| `raise`            | Re-throw the current exception   |
