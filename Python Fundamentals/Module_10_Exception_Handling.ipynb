# Exception Handling

##### What is an Exception?

**An exception is an event that occurs during program execution which interrupts the normal flow of the program.**

**Common Python Exception**

| Exception           | Cause                     |
| ------------------- | ------------------------- |
| `ZeroDivisionError` | Divide by zero            |
| `ValueError`        | Invalid value             |
| `TypeError`         | Wrong data type           |
| `IndexError`        | Invalid list index        |
| `KeyError`          | Missing dictionary key    |
| `AttributeError`    | Object lacks an attribute |
| `FileNotFoundError` | File does not exist       |
| `NameError`         | Variable not defined      |

**try**

The try block contains code that might raise an exception.

**except**

Handles a specific exception

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Division by zero is not allowed.")
```

```python
try:
    value = int(input())
    print(10 / value)

except ValueError:
    print("Please enter a valid integer.")

except ZeroDivisionError:
    print("Cannot divide by zero.")
```

**else**

The else block executes only if no exception occurs.

```python
try:
    print(10 / 2)

except ZeroDivisionError:
    print("Error")

else:
    print("Everything succeeded")
```

**finally**

This block always execute

```python
try:
    print(10 / 0)

except ZeroDivisionError:
    print("Error")

finally:
    print("Cleanup")
```
