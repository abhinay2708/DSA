# Strings

### What is a String?

**A string is an immutable sequence of Unicode characters.**

- Immutable → cannot be modified after creation.
- Sequence → supports indexing, slicing, iteration.
- Unicode → supports characters from almost every language.

**Creating Strings**

```python
# Single quotes
name = 'Abhinay'

# Double Quotes
city = "Hyderabad"

# Triple Quotes
text = """I
am a
Data
Scientist"""

print(name,city, text)
```

**String Indexing**

```python
word = "Python"
word[0]
```

**Negative Indexing**

```python
word[-1]
```

**Slicing**

syntax--string[start:stop:step]

```python
word = "Python"

print(word[1:4])
print(word[:4])
print(word[2:])
print(word[:])
print(word[::-1])
print(word[::2])
print(word[5:2:-1])
```

**String are immutable**

```python
word = "Python"

word[0] = "J"
```

```python
word = "J" + word[1:]
print(word)
```

**String Length**

```python
len(word)
```

**Membership Operator**

```python
"Py" in word
```

**String Traversal**

```python
for i in word:
    print(i, end=" ")
print()
for i in range(len(word)):
    print(word[i],end=" ")
```
