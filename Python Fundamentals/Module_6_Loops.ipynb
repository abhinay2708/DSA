# What is Loop?

**A loop is a control flow statement that repeatedly executes a block of code until a condition becomes false or all elements in an iterable have been processed.**

**Python has two loop types:**

- while
- for

1. While Loop

**A while loop repeats as long as the condition is True.**

```python
i = 1

while i <= 5:
    print(i, end=" ")
    i += 1
```

```python
# Infinite Loop

while True:
    print("Hello")
    
# This never stops unless you explicitly break or terminate the program.
```

2. For loop

**Python's for loop iterates over an iterable.**

```python
for i in [10,20,30]:
    print(i, end=" ")
```

##### range()

1. range(stop)

```python
for i in range(5):
    print(i)

# Rule: Starts from 0 and stops before 5
```

2. range(start,stop)

- start → included
- stop → excluded

```python
for i in range(2,6):
    print(i)
```

3. range(start,stop,step)

```python
for i in range(0,10,2):
    print(i)
```

4. Negative step

```python
for i in range(10,0,-2):
    print(i)
```

**Iterating Over Different Iterables**

```python
# List

nums = [10, 20, 30]

for num in nums:
    print(num)
```

```python
# String

for ch in "Python":
    print(ch)
```

```python
# Dictionary

person = {
    "name": "Abhinay",
    "age": 25
}

for key, value in person.items():
    print(key, value)
```

**break**

stops the loop immidiately

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

**continue**

Skips the current iterations

```python
for i in range(5):
    if i == 2:
        continue
    print(i)
```

**pass**

Does nothing, used as a placeholder

```python
if True:
    pass
```

**for-else**

The else block of a loop executes only if the loop completes normally without encountering a break

```python
for i in range(5):
    print(i)
else:
    print("Finished")
```

```python
for i in range(5):
    if i == 3:
        break
    print(i)
else:
    print("Finished")
```

##### Nested Loops

```python
for i in range(3):
    for j in range(2):
        print(i, j)
```
